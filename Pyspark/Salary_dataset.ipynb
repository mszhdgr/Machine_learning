```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Salary_ds").getOrCreate()
spark
```

```python
df = spark.read.csv("/content/MOCK_DATA.csv", header=True, inferSchema=True)
df.show(5)
```

```python
df.printSchema()
```

```python
df.summary().show()
```

```python
df.select(["first_name","last_name","Age"]).show(5)
```

```python
df.filter((df.Age > 20) & (df.Age < 25)).show(5)
```

```python
df.where((df.Age < 30) & (df.Salary > 6000)).show(5)
```

```python
df = df.withColumn("TotalSalary", df.Salary + df.Incentives)
df.show(5)
```

```python
df = df.withColumnRenamed("first_name", "FirstName")
df = df.withColumnRenamed("last_name", "LastName")
df.show(5)
```

```python
from pyspark.sql import functions as F

df.groupBy(df.Department).agg(
    F.min("Salary").alias("Min salary"),
    F.max("Salary").alias("Max salary")
).show()
```

```python
df.createTempView("Salary_Dataset")
```

```python
spark.sql(
    "SELECT * FROM Salary_Dataset"
    ).show()
```

```python
spark.sql(
    "SELECT FirstName, Salary FROM Salary_Dataset LIMIT 6"
    ).show()
```

```python
spark.sql(
    "SELECT DISTINCT(Department) FROM Salary_Dataset"
    ).show()
```
