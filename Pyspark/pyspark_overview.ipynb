```python
from pyspark.sql import SparkSession
from pyspark.sql.types import *
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

```python
spark = SparkSession.builder.appName("Pyspark cheatsheet").getOrCreate()
```

```python
spark
```

```python
spark.sparkContext.setLogLevel("WARN")
```

```python
data = [
    (1, "john", "doe", 30, "Engineering", 75000, "2020-01-15"),
    (2, "Jane", "Smith", 25, "Marketing", 65000, "2021-03-20"),
    (3, "Bob", "Johnson", 35, "Engineering", 85000, "2019-07-10"),
    (4, "Alice", "Williams", 28, "Sales", 70000, "2020-11-05"),
    (5, "Charlie", "Brown", 40, "Marketing", 95000, "2018-09-25"),
    (6, "Diana", "Jones", 32, "Engineering", 78000, "2021-06-15"),
    (7, "Eve", "Miller", 27, "Sales", 62000, "2022-02-01"),
    (8, "Frank", "Davis", 45, "Management", 110000, "2017-12-01"),
    (9, "Grace", "Garcia", 29, "Engineering", 72000, "2021-08-20"),
    (10, "Henry", "Rodriguez", 38, "Marketing", 88000, "2019-04-12")
]
data
```

```python
schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("first_name", StringType(), True),
    StructField("last_name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("department", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("hire_date", StringType(), True),
])
schema
```

```python
df = spark.createDataFrame(data, schema)
df.show(5)
```

```python
df.withColumn("salary_category",
              when(col("salary") >= 100000, "High") \
              .when(col("salary") >= 75000, "Medium") \
              .otherwise("Low")
              ).show()
```

```python
z = [
    (1, "John", "Doe", 30, "Engineering", 75000, "2020-01-15"),
    (2, "Jane", "Smith", 25, "Marketing", 65000, "2021-03-20"),
    (3, "Bob", "Johnson",None, "Engineering",None, "2019-07-10"),
    (4, "Alice", "Williams", 28, "Sales", 70000, "2020-11-05"),
    (5, "Charlie", "Brown", 40, "Marketing", 95000, "2018-09-25"),
    (6, "Diana", "Jones", 32, "Engineering", 78000, "2021-06-15"),
    (7, "Eve", "Miller", 27, "Sales", None, "2022-02-01"),
    (8, "Frank", "Davis", 45, "Management", 110000, "2017-12-01"),
    (9, "Grace", "Garcia", None, "Engineering", 72000, "2021-08-20"),
    (10, "Henry", "Rodriguez", 38, "Marketing", 88000, "2019-04-12")
]
df2 = spark.createDataFrame(z, ["id", "first_name", "last_name", "age",
                                   "department", "salary", "hire_date"])
```

```python
df2.show()
```

```python
df2.withColumn("salary_filled", coalesce(col("salary"), lit(0))).show()
```

```python
df2.na.fill({"age": 20, "salary": 60000}).show()
```

```python
df2.na.drop(how="any").show()
```

```python
df2.na.drop(how="all").show()
```

```python
df2.select(
    count(col("id")).alias("total_employees"),
    avg(col("salary")).alias("average_salary"),
    max(col("salary")).alias("maximum_salary"),
    min(col("salary")).alias("minimum_salary"),
).show()
```

```python
df2.groupBy("department").agg(
    count(col("salary")).alias("total_employees"),
    sum(col("salary")).alias("total_salary"),
    min(col("salary")).alias("minimum_salary"),
    max(col("salary")).alias("max_salary"),
).show()
```

```python
df2.groupBy("department").agg(
    collect_list("first_name").alias("employees"),
    collect_set("first_name").alias("unique_employees"),
).show()
```

```python
df2.groupBy("department").pivot("age").count().show()
```

```python
df2.rollup("department", "age").agg(sum("salary")).show()
```

```python
dept_data = [
    ("Engineering", "Building_A", "07-99"),
    ("Sales", "Building_B", "01-89"),
    ("Marketing", "Building_D", "07-23"),
    ("Management", "Building_C", "06-59")
]
dept_df = spark.createDataFrame(dept_data, ["department", "building", "phone"])
dept_df.show()
```

```python
df2.join(dept_df, df2.department == dept_df.department, "inner").show()
```

```python
df2.join(dept_df, df2.department == dept_df.department, "left").show()
```

```python
df2.join(dept_df, df2.department == dept_df.department, "right").show()
```

```python
joined_df = df2.join(dept_df, df2.department == dept_df.department, "inner")
joined_df.show()
```

```python
joined_df.select(
    col("first_name").alias("name"),
    col("salary"),
    col("building"),
    col("phone")
).show()
```

```python
window_spec = Window.partitionBy("department").orderBy(col("salary").desc())
```

```python
df2.withColumn("rank", rank().over(window_spec)).show()
```

```python
df2.withColumn("rank", rank().over(window_spec)).filter(col("rank") == 1).show()
```

```python
df2.withColumn("row_numb", row_number().over(window_spec)).show()
```
