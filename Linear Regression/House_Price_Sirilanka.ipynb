```python
import warnings

warnings.filterwarnings("ignore")
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("nishanobeysekera/house-prices")

print("Path to dataset files:", path)
```

```python
import pandas as pd

df = pd.read_csv("/kaggle/input/house-prices/house_prices_srilanka.csv")
df.head()
```

```python
df.info()
```

```python
df['Location'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
encoded_location = encoder.fit_transform(df['Location'])
```

```python
encoded_location
```

```python
df['Encoded_Location'] = encoded_location
df.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
sf_encoded = scaler.fit_transform(df['SquareFootage'].values.reshape(-1,1))
df['sq_encoded'] = sf_encoded
df.head()
```

```python
featurez = ['Bedrooms','sq_encoded','Encoded_Location']
labels = ['Price']
x = pd.DataFrame(df, columns=featurez)
y = pd.DataFrame(df, columns=labels)
x.head()
```

```python
y.tail()
```

```python
y = y.values.ravel()
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
len(X_train), len(X_test)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python
from sklearn.metrics import r2_score

pred = model.predict(X_test)
r2 = r2_score(y_test, pred)
r2
```

```python
model.feature_names_in_
```

```python
test_in = [[1,700,0],[2, 1600, 1], [5, 4800, 2]]
test_df = pd.DataFrame(test_in, columns=model.feature_names_in_)
test_df
```

```python
test_df['sq_encoded'] = scaler.transform(test_df[['sq_encoded']])
test_df
```

```python
predictions = model.predict(test_df)
```

```python
predictions
```
