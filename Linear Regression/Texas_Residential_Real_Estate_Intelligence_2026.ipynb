```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("jahnavikachhia23/texas-residential-real-estate-intelligence-2026")

print("Path to dataset files:", path)
```

```python
import pandas as pd

df = pd.read_csv("/kaggle/input/texas-residential-real-estate-intelligence-2026/Texas_Residential_Real Estate_Intelligenc_ 2026.csv")
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df.head(3)
```

```python
features = df.drop(columns=['type','sub_type','text','listPrice'])
features.head()
```

```python
label = pd.DataFrame(df, columns=['listPrice'])
label
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
scaled_features
```

```python
y = label.values.ravel()
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(scaled_features, y, test_size=0.2)
len(X_train), len(X_test)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
accuracy = model.score(X_test,y_test)
accuracy
```

```python
from sklearn.metrics import r2_score

pred = model.predict(X_test)
r2 = r2_score(y_test, pred)
r2
```
