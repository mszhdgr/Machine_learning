## Area of a Circle

- formulae is pi * r^2

```python
PI = 3.14
```

```python
PI
```

```python
import numpy as np

r = np.array(range(0, 1000, 2)).reshape(-1, 1)
r[:5]
```

```python
len(r)
```

```python
np.pow(2, 2)
```

```python
squares = [np.pow(i, 2).item() for i in r]
squares[:5]
```

```python
y = [PI * i for i in squares]
y[:5]
```

```python
type(y)
```

```python
y = np.array(y)
type(y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test= train_test_split(r, y, test_size=0.3)
```

```python
len(X_train), len(X_test)
```

```python
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

preds = model.predict(X_test)
R2 = r2_score(y_test, preds)
MAE = mean_absolute_error(y_test, preds)
MSE = mean_squared_error(y_test, preds)

print(f"R2: {R2}")
print(f"MAE: {MAE}")
print(f"MSE: {MSE}")
```

## Deep learning regression recap

```python
import torch
torch.__version__
```

```python
from torch import nn
```

```python
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)
    def forward(self, x):
        return self.linear(x)
```

```python
model1 = SimpleNN()
model1
```

```python
x = torch.arange(0, 100, 0.02).reshape(-1, 1)
m = 0.7
c = 0.3
y = m * x + c
```

```python
X_train, X_test, y_train, y_test= x[:400], x[400:], y[:400], y[400:]
```

```python
from torch.nn import L1Loss
from torch.optim import SGD

criterion = L1Loss()
optimizer = SGD(params=model_1.parameters(), lr=0.1)
```

```python
for epoch in range(1000):
    model_1.train()
    pred = model_1(X_train.float())
    loss = criterion(y_train, pred)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 100 == 0:
        print(f"__Epoch__: {epoch}\n")
        print(f"Loss__: {loss:.5f}")
```
