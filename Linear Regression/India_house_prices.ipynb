```python
import warnings

warnings.filterwarnings("ignore")
```

```python
import kagglehub

path = kagglehub.dataset_download("sukhmandeepsinghbrar/house-prices-india")

print("Path to dataset files:", path)
```

```python
import pandas as pd

df = pd.read_csv("/kaggle/input/house-prices-india/House Price India.csv")
df.head()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
columns = ['number of bedrooms', 'number of bathrooms','condition of the house','Number of schools nearby', 'Distance from the airport']
features = pd.DataFrame(df, columns=columns)
features.head()
```

```python
features.describe()
```

```python
labels = pd.DataFrame(df, columns=['Price'])
labels.head()
```

```python
len(features)
```

```python
y = labels.values.ravel()
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(features, y, test_size=0.3)
```

```python
len(X_train), len(X_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.decomposition import  PCA
```

```python
ct = Pipeline(
    steps=[
        ('pca', PCA(2)),
        ('lr_model', LinearRegression())
    ]
)
```

```python
ct.fit(X_train, y_train)
```

```python
ct.named_steps
```

```python
ct.score(X_test,y_test)
```

```python
ct.feature_names_in_
```

```python
import numpy as np

valz = np.array([1, 1, 2, 1, 300])
pred = ct.predict(valz.reshape(1, -1))
pred
```

```python
valz2 = np.array([4, 5, 4, 4, 70])
pred2 = ct.predict(valz2.reshape(1, -1))
pred2
```
