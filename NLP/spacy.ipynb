### Custom Text Classifier

```python
TRAIN_DATA = [("Konnect has installed Wifi in Roysambu",
              {"entities": [(0, 7, "NETINSTALLER"), (30,38, "CONSTITUENCY")]})]
TRAIN_DATA
```

```python
import spacy
from spacy.training.example import Example

nlp = spacy.blank("en")
```

```python
ner = nlp.add_pipe("ner")
```

```python
ner.add_label("NETINSTALLER")
ner.add_label("CONSTITUENCY")
```

```python
optimizer = nlp.begin_training()
```

```python
import random

for i in range(20):
    random.shuffle(TRAIN_DATA)
    losses = {}
    for text, annotations in TRAIN_DATA:
        doc = nlp.make_doc(text)
        example = Example.from_dict(doc, annotations)
        nlp.update([example], sgd=optimizer, drop=0.35, losses=losses)
        print(losses)
        
nlp.to_disk("CustomNERv1")
```

```python
nlp = spacy.load("CustomNERv1")
```

```python
doc = nlp("Konnect wifi is Down All over Roysambu")
```

```python
displacy.render(doc, style="ent")
```
