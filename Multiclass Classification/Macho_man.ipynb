```python
from sklearn.datasets import load_iris

data = load_iris()
x, y = data.data, data.target
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y)
len(X_train), len(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

def Model_Train(model,xtrain, ytrain, xtest):
  model.fit(xtrain, ytrain)
  pred = model.predict(xtest)
  cr = classification_report(y_test, pred)
  print(cr)
  cm = confusion_matrix(y_test, pred)
  print(cm)
  ConfusionMatrixDisplay(cm).plot()
  plt.show()
```

### Support Vector Machines

```python
Model_Train(SVC(), X_train, y_train, X_test)
```

### Multinomial Naive Bayes

```python
Model_Train(MultinomialNB(), X_train, y_train, X_test)
```

### Random Forest Classifier

```python
Model_Train(RandomForestClassifier(), X_train, y_train, X_test)
```

### Xgboost Classifier

```python
Model_Train(XGBClassifier(), X_train, y_train, X_test)
```

KNN Classifier

```python
Model_Train(KNeighborsClassifier(n_neighbors=3), X_train, y_train, X_test)
```
