```python
import warnings

warnings.filterwarnings("ignore")
```

```python
import kagglehub

path = kagglehub.dataset_download("harshitshankhdhar/imdb-dataset-of-top-1000-movies-and-tv-shows")

print("Path to dataset files:", path)
```

```python
import pandas as pd

df = pd.read_csv("/kaggle/input/imdb-dataset-of-top-1000-movies-and-tv-shows/imdb_top_1000.csv")
df.head()
```

```python
df.columns
```

```python
data = pd.DataFrame(df, columns=['IMDB_Rating','Genre','Series_Title','Overview'])
data.head()
```

```python
data.isna().sum()
```

```python
data['Overview']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfid = TfidfVectorizer()
tfid_vect = tfid.fit_transform(data['Overview'])
```

```python
tfid_vect
```

```python
tfid.get_feature_names_out()
```

```python
tfid_vect.toarray()
```

```python
df.index
```

```python
indices = pd.Series(df.index, df['Series_Title'])
indices
```

```python
data[data['Series_Title'] == 'Giant']
```

```python
indices['Giant']
```

```python
from sklearn.metrics.pairwise import cosine_similarity

cos_sim = cosine_similarity(tfid_vect, tfid_vect)
```

```python
cos_sim
```

```python
movie_idx = indices['The Godfather']
movie_idx
```

```python
sim_scores = cos_sim[movie_idx]
sim_scores[:5]
```

```python
similarity_scores = pd.DataFrame(sim_scores, columns=['Scores'])
similarity_scores
```

```python
movie_indexes = similarity_scores.sort_values('Scores', ascending=False)[1:11].index
movie_indexes
```

```python
data['Series_Title'].iloc[movie_indexes]
```

```python
mv_idx = indices['Gladiator']
mv_idx
```

```python
glad_simscore = cos_sim[mv_idx]
glad_simscore[:5]
```

```python
glad_sim_score = pd.DataFrame(glad_simscore, columns=['Scores'])
glad_sim_score
```

```python
gladiator_idx = glad_sim_score.sort_values('Scores', ascending=False)[1:11].index
gladiator_idx
```

```python
data['Series_Title'][gladiator_idx]
```
