# Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# regression formula -> y = mx + c

weight = 0.7
bias = 0.3

x = torch.arange(0,1,0.02).reshape(-1, 1)
y = weight * x + bias
```

```python
x[:5], y[:5]
```

```python
len(x), len(y)
```

```python
x.shape, x.ndim
```

```python

X_train = x[:40]
X_test = x[40:]
y_train = y[:40]
y_test = y[40:]
```

```python
len(X_train), len(X_test)
```

```python
X_train[0], y_train[0]
```

```python
from torch import nn
```

```python
class LinearModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)
    def forward(self, x):
        return self.linear(x)
```

```python
model = LinearModel()
```

```python
model
```

```python
with torch.inference_mode():
    pred = model(X_train[0])

pred
```

```python
from torch.nn import L1Loss
from torch.optim import SGD
```

```python
criterion = L1Loss()
optimizer = SGD(params=model.parameters(), lr=0.01)
```

```python
from tqdm import tqdm
```

```python
epochs = 1000

for epoch in tqdm(range(epochs)):
    model.train()
    pred = model(X_train)
    loss = criterion(pred, y_train)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 100 == 0:
        print(f"__Epoch__:{epoch}\n")
        print(f"LOSS:{loss:.5f}\n")
```
